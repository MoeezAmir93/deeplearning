```python
# Total model parameters
total_parameters = 175e9  # 175 Billion

# Attention parameter calculation
embedding_dim = 12288
kqv_dim = 128
num_heads = 96
num_layers = 96
matrices_per_head = 4  # Key, Query, Value, Output

# Parameters per matrix (for a single projection)
parameters_per_matrix = embedding_dim * kqv_dim  # 128 * 128
parameters_per_head = parameters_per_matrix * matrices_per_head
parameters_per_block = parameters_per_head * num_heads
total_attention_parameters = parameters_per_block * num_layers
attention_parameters_billion = total_attention_parameters / 1e9

# Feedforward network in transformer block
# Typically 2 fully connected layers: hidden_size -> 4*hidden_size -> hidden_size
# So FFN params per layer ≈ 2 × hidden_dim × (4 × hidden_dim) = 8 × hidden_dim²
ffn_parameters_per_layer = 8 * embedding_dim ** 2
total_ffn_parameters = ffn_parameters_per_layer * num_layers
ffn_parameters_billion = total_ffn_parameters / 1e9

# Embedding layer parameters: vocab_size × embedding_dim
vocab_size = 50000  # common size for GPT
embedding_parameters = vocab_size * embedding_dim
embedding_parameters_billion = embedding_parameters / 1e9

# LayerNorms and biases (approx): estimate as ~1% of total parameters
layernorm_bias_parameters = total_parameters * 0.01
layernorm_bias_billion = layernorm_bias_parameters / 1e9

# Sum all computed components
estimated_total = (total_attention_parameters + total_ffn_parameters +
                   embedding_parameters + layernorm_bias_parameters)
estimated_total_billion = estimated_total / 1e9

# Print breakdown
print("===== PARAMETER BREAKDOWN =====")
print(f"Total Parameters: {total_parameters/1e9:.0f} Billion")
print(f"-> Attention Parameters: {attention_parameters_billion:.2f} Billion")
print(f"-> FFN Parameters: {ffn_parameters_billion:.2f} Billion")
print(f"-> Embedding Parameters: {embedding_parameters_billion:.2f} Billion")
print(f"-> LayerNorm/Bias: {layernorm_bias_billion:.2f} Billion")
print("--------------------------------")
print(f"Estimated Total: {estimated_total_billion:.2f} Billion")
print(f"Difference from 175B: {total_parameters/1e9 - estimated_total_billion:.2f} Billion")
```
