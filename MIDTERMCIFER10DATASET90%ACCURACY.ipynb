```python
# Importing Required Libraries
import numpy as np
from tensorflow.keras import layers, models, optimizers, utils, datasets
import matplotlib.pyplot as plt

# 1. Data Loading and Preprocessing
NUM_CLASSES = 10

# Load CIFAR-10 dataset
(x_train, y_train), (x_test, y_test) = datasets.cifar10.load_data()

# Normalize the data to the range [0, 1]
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Convert labels to one-hot encoding
y_train = utils.to_categorical(y_train, NUM_CLASSES)
y_test = utils.to_categorical(y_test, NUM_CLASSES)

# 2. Building the CNN Model

input_layer = layers.Input((32, 32, 3))

# First Convolution Block
x = layers.Conv2D(filters=32, kernel_size=3, strides=1, padding="same")(input_layer)
x = layers.BatchNormalization()(x)
x = layers.LeakyReLU()(x)

# Second Convolution Block
x = layers.Conv2D(filters=32, kernel_size=3, strides=2, padding="same")(x)
x = layers.BatchNormalization()(x)
x = layers.LeakyReLU()(x)

# Third Convolution Block
x = layers.Conv2D(filters=64, kernel_size=3, strides=1, padding="same")(x)
x = layers.BatchNormalization()(x)
x = layers.LeakyReLU()(x)

# Fourth Convolution Block
x = layers.Conv2D(filters=64, kernel_size=3, strides=2, padding="same")(x)
x = layers.BatchNormalization()(x)
x = layers.LeakyReLU()(x)

# Flatten the output of the convolution layers
x = layers.Flatten()(x)

# Fully Connected Layer with Dropout
x = layers.Dense(128)(x)
x = layers.BatchNormalization()(x)
x = layers.LeakyReLU()(x)
x = layers.Dropout(rate=0.5)(x)

# Output Layer with Softmax activation
x = layers.Dense(NUM_CLASSES)(x)
output_layer = layers.Activation("softmax")(x)

# Create the Model
model = models.Model(input_layer, output_layer)

# Print the model summary
model.summary()

# 3. Compile the Model

# Set the optimizer and compile the model
opt = optimizers.Adam(learning_rate=0.0005)
model.compile(loss="categorical_crossentropy", optimizer=opt, metrics=["accuracy"])

# 4. Train the Model

# Train the model on the training data
model.fit(
    x_train,
    y_train,
    batch_size=32,
    epochs=10,
    shuffle=True,
    validation_data=(x_test, y_test),
)

# 5. Evaluate the Model

# Evaluate the model on the test data
test_loss, test_accuracy = model.evaluate(x_test, y_test, batch_size=1000)
print(f"Test accuracy: {test_accuracy:.4f}, Test loss: {test_loss:.4f}")

# 6. Make Predictions and Display Results

# List of class labels for CIFAR-10 dataset
CLASSES = np.array(
    [
        "airplane", "automobile", "bird", "cat", "deer", "dog", "frog",
        "horse", "ship", "truck",
    ]
)

# Make predictions on the test set
preds = model.predict(x_test)
preds_single = CLASSES[np.argmax(preds, axis=-1)]
actual_single = CLASSES[np.argmax(y_test, axis=-1)]

# 7. Display Predictions vs Actual Labels

# Set up for displaying random test images with predictions
n_to_show = 10
indices = np.random.choice(range(len(x_test)), n_to_show)

# Create a figure for displaying the images
fig = plt.figure(figsize=(15, 3))
fig.subplots_adjust(hspace=0.4, wspace=0.4)

# Loop through the selected images and display them with predictions
for i, idx in enumerate(indices):
    img = x_test[idx]
    ax = fig.add_subplot(1, n_to_show, i + 1)
    ax.axis("off")
    ax.text(
        0.5, -0.35, "pred = " + str(preds_single[idx]), fontsize=10, ha="center", transform=ax.transAxes
    )
    ax.text(
        0.5, -0.7, "act = " + str(actual_single[idx]), fontsize=10, ha="center", transform=ax.transAxes
    )
    ax.imshow(img)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers, callbacks
from tensorflow.keras.applications import DenseNet121
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Constants
NUM_CLASSES = 10  # CIFAR-10 has 10 classes
IMG_SIZE = (64, 64)  # Resize to 64x64 for DenseNet

# Load CIFAR-10 dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

# Preprocess the images
x_train = tf.cast(x_train, tf.float32) / 255.0
x_test = tf.cast(x_test, tf.float32) / 255.0

# Resize images to 64x64
x_train = tf.image.resize(x_train, IMG_SIZE)
x_test = tf.image.resize(x_test, IMG_SIZE)

# One-hot encode the labels
y_train = tf.keras.utils.to_categorical(y_train, NUM_CLASSES)
y_test = tf.keras.utils.to_categorical(y_test, NUM_CLASSES)

# Initialize DenseNet121 model with pre-trained weights
base_model = DenseNet121(weights='imagenet', include_top=False, input_shape=(64, 64, 3))

# Build the model
model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(512, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(NUM_CLASSES, activation='softmax')
])

# Freeze the base model layers initially
base_model.trainable = False

# Compile the model
model.compile(optimizer=optimizers.Adam(learning_rate=0.0005),
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Set up data augmentation
datagen = ImageDataGenerator(
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)

# Train the model (initial training with base model frozen)
history = model.fit(
    datagen.flow(x_train, y_train, batch_size=64),
    steps_per_epoch=len(x_train) // 64,
    epochs=20,
    validation_data=(x_test, y_test),
    callbacks=[callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)]
)

# Fine-tune the DenseNet-121 layers
base_model.trainable = True  # Unfreeze all layers
for layer in base_model.layers[:100]:  # Optionally unfreeze the top layers first
    layer.trainable = False

# Recompile the model with a lower learning rate
model.compile(optimizer=optimizers.Adam(learning_rate=0.0001),
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Continue training (fine-tuning the DenseNet layers)
history_fine_tune = model.fit(
    datagen.flow(x_train, y_train, batch_size=64),
    steps_per_epoch=len(x_train) // 64,
    epochs=20,
    validation_data=(x_test, y_test),
    callbacks=[callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)]
)

# Evaluate the model on test data
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f"Test accuracy after fine-tuning: {test_acc * 100:.2f}%")
```
